# Data Cleaning 

```python
# Drop the first column
df.drop(df.columns[0], axis=1, inplace=True)
```

```python
# Remove Missing Values
df.dropna(inplace=True)
```

```python
# Check Again
df.isnull().sum()
```

```python
# Reset Index
df.reset_index(drop=True, inplace=True)
```

```python
# Convert Date
df["Date"] = pd.to_datetime(df["Date"], format="%y/%m/%d")
```

# Feature Engineering

```python
# Month
df['Month'] = df['Date'].dt.month_name()
```

```python
# Day
df['Day'] = df['Date'].dt.day_name()
```

```python
# Year
df['Year'] = df['Date'].dt.year
```

```python
# Average Sales Per Item
df['Sales_Per_Item'] = df['Amount'] / df['Qty']
```

# EDA

### Q1 : Which product category has the highest number of orders ?

```python
category_orders = (
    df.groupby("Category")["Order ID"]
    .count()
    .reset_index()
)

fig = px.histogram(df, x='Category', 
                   title="Number of Orders by Category",
                  )

fig.show()
```

### Q2 : Which product size is ordered the most ?

```python
fig = px.histogram(
    df,
    x='Size',
    category_orders={
        'Size': df['Size'].value_counts().index.tolist()
    }
)

fig.show()
```

### Q3 : Which category generates the highest revenue?

```python
category_sales = (
    df.groupby("Category")["Amount"]
    .sum()
    .reset_index()
)

fig = px.bar(
    category_sales,
    x="Category",
    y="Amount",
    color="Category",
    title="Total Sales by Category",
    category_orders={
        "Category": category_sales.sort_values(
            "Amount", ascending=False
        )["Category"].tolist()
    }
)

fig.show()
```

### Q4 : Which cities have the highest number of orders?

```python
city_orders = (
    df['ship-city']
    .value_counts()
    .head(10)
    .reset_index()
)

city_orders.columns = ['City', 'Number of Orders']

fig = px.bar(
    city_orders,
    x='City',
    y='Number of Orders',
    color='Number of Orders',
    text='Number of Orders',
    title='Top 10 Cities by Number of Orders'
)

fig.update_layout(xaxis={'categoryorder': 'total descending'})

fig.show()
```

### Q5 : Which month has the highest sales?

```python
monthly_sales = (
    df.groupby("Month")["Amount"]
    .sum()
    .reset_index()
)

month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

monthly_sales["Month"] = pd.Categorical(
    monthly_sales["Month"],
    categories=month_order,
    ordered=True
)

monthly_sales = monthly_sales.sort_values("Month")

fig = px.line(
    monthly_sales,
    x="Month",
    y="Amount",
    markers=True,
    title="Monthly Sales"
)

fig.show()
```

### Q6 : Which order status appears most?

```python
fig = px.histogram(
    df,
    x="Status",
    color="Status",
    title="Order Status Distribution",
    category_orders={
        "Status": df["Status"].value_counts().index.tolist()
    }
)

fig.show()
```
